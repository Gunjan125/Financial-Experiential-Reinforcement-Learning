```python
import yfinance as yf
import faiss
import re
import json
import torch
import gc
from datetime import datetime
from typing import Dict, List, Tuple
from sentence_transformers import SentenceTransformer
from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM

class ReflectionMemory:
    def __init__(self, embedder):
        self.embedder = embedder
        self.dim = 384
        self.index = faiss.IndexFlatL2(self.dim)
        self.reflections: List[str] = []

    def add(self, delta: str):
        emb = self.embedder.encode([delta], convert_to_numpy=True)
        self.index.add(emb)
        self.reflections.append(delta)

    def get_relevant(self, query: str, top_k: int = 2) -> str:
        if len(self.reflections) == 0:
            return ""
        q_emb = self.embedder.encode([query], convert_to_numpy=True)
        _, I = self.index.search(q_emb, top_k)
        return " ".join([self.reflections[i] for i in I[0] if i < len(self.reflections)])

class TradingEnv:
    def __init__(self, df):
        self.df = df.reset_index(drop=True)
        self.initial_balance = 10000.0
        self.reset()

    def reset(self):
        self.current_step = 0
        self.balance = self.initial_balance
        self.shares = 0.0
        return self._get_state()

    def _get_state(self):
        row = self.df.iloc[self.current_step]
        return {
            'news': str(row['Combined_News'])[:300],
            'price': float(row['DJIA_Close']),
            'portfolio_value': self.balance + self.shares * float(row['DJIA_Close'])
        }

    def simulate(self, action: Dict) -> Tuple[Dict, float, str]:
        price = float(self.df.iloc[self.current_step]['DJIA_Close'])
        frac = min(max(action.get('fraction', 0.5), 0.0), 1.0)

        temp_balance = self.balance
        temp_shares = self.shares

        if action.get('decision') == 'buy' and temp_balance > 0:
            invest = temp_balance * frac
            temp_shares += invest / price
            temp_balance -= invest
        elif action.get('decision') == 'sell' and temp_shares > 0:
            sell = temp_shares * frac
            temp_balance += sell * price
            temp_shares -= sell

        next_idx = self.current_step + 1
        next_price = float(self.df.iloc[next_idx]['DJIA_Close']) if next_idx < len(self.df) else price

        next_value = temp_balance + temp_shares * next_price
        current_value = self.balance + self.shares * price
        reward = (next_value - current_value) / current_value if current_value > 0 else 0.0

        feedback = (f"Action: {action['decision']} {frac*100:.1f}% | "
                    f"Next-day change: {(next_price/price-1)*100:+.2f}% | "
                    f"Decision return: {reward*100:+.2f}%")

        next_state = {**self._get_state(), 'portfolio_value': next_value}
        return next_state, reward, feedback

    def step(self, action: Dict):
        
        
        price = float(self.df.iloc[self.current_step]['DJIA_Close'])
        frac = min(max(action.get('fraction', 0.5), 0.0), 1.0)

        if action.get('decision') == 'buy' and self.balance > 0:
            invest = self.balance * frac
            self.shares += invest / price
            self.balance -= invest
        elif action.get('decision') == 'sell' and self.shares > 0:
            sell = self.shares * frac
            self.balance += sell * price
            self.shares -= sell

        self.current_step += 1
        done = self.current_step >= len(self.df) - 1
        next_price = float(self.df.iloc[self.current_step]['DJIA_Close']) if not done else price
        next_value = self.balance + self.shares * next_price
        reward = (next_value - self.balance - self.shares * price) / (self.balance + self.shares * price) if (self.balance + self.shares * price) > 0 else 0.0
        feedback = f"Real step executed. Portfolio return this day: {reward*100:+.2f}%"
        return self._get_state(), reward, done, feedback

# ====================== ERL AGENT ======================
class ERLAgent:
    def __init__(self, generator, embedder, tokenizer):
        self.generator = generator
        self.embedder = embedder
        self.tokenizer = tokenizer
        self.memory = ReflectionMemory(embedder)
        self.internalized: List[Tuple[str, Dict]] = []
        self.tau = 0.0005

    def _format_x(self, state: Dict) -> str:
        return f"News: {state['news']}\nPrice: {state['price']:.2f}\nPortfolio: {state['portfolio_value']:.2f}"
    def _generate(self, user_prompt: str, max_new_tokens=80, do_sample=True, temperature=0.7):
            messages = [
                {"role": "system", "content": "You are an expert trading agent. Always respond with valid JSON only."},
                {"role": "user", "content": user_prompt}
            ]
            formatted = self.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
            
            out = self.generator(
                formatted,
                max_new_tokens=max_new_tokens,
                do_sample=do_sample,
                temperature=temperature,
                top_p=0.9,
                pad_token_id=self.tokenizer.eos_token_id
            )[0]['generated_text']
            return out[len(formatted):].strip()
    def decide(self, state: Dict, reflection: str = "") -> Dict:
        x_str = self._format_x(state)
        fewshot = ""
        if self.internalized:
            fewshot = "Learned successful trades:\n" + \
                      "\n".join([f"X: {s}\nAction: {json.dumps(a)}" for s, a in self.internalized[-5:]]) + "\n\n"

        refl_text = f"Previous reflection to apply: {reflection}\n\n" if reflection else ""

        prompt = f"""{fewshot}{refl_text}Current market situation:
{x_str}
Output ONLY valid JSON:
{{"decision": "buy"|"sell"|"hold", "fraction": 0.0-1.0}}"""

        resp = self._generate(prompt, max_new_tokens=80, do_sample=True, temperature=0.7)

        try:
            match = re.search(r'\{.*?\}', resp, re.DOTALL)
            action = json.loads(match.group(0)) if match else {}
            action['decision'] = action.get('decision', 'hold').lower()
            action['fraction'] = float(action.get('fraction', 0.5))
            return action
        except:
            return {"decision": "hold", "fraction": 0.0}

    def generate_reflection(self, x_str: str, y1: Dict, f1: str, r1: float) -> str:
        mem_context = self.memory.get_relevant(x_str, top_k=2)
        prompt = f"""Past successful lessons: {mem_context}

Task: {x_str}
First attempt: {y1}
Feedback: {f1}
Reward: {r1:.4f}

Write a short 1-2 sentence reflection: what was good/bad and exactly how to improve the decision for this exact situation.

Reflection:"""

        return self._generate(prompt, max_new_tokens=120, do_sample=False)

    def internalize(self, x_str: str, y2: Dict):
        summary = x_str[:250]
        self.internalized.append((summary, y2))
        if len(self.internalized) > 15:
            self.internalized.pop(0)

def run_erl(env: TradingEnv, agent: ERLAgent, num_days: int = 100):
    env.reset()
    returns = []
    for day in range(num_days):
        if env.current_step >= len(env.df) - 2:
            break
        state = env._get_state()
        x_str = agent._format_x(state)

        y1 = agent.decide(state, "")
        _, r1, f1 = env.simulate(y1)

        delta = agent.generate_reflection(x_str, y1, f1, r1)

        y2 = agent.decide(state, delta)
        _, r2, f2 = env.simulate(y2)
        if r2 > agent.tau:
            agent.memory.add(delta)
            agent.internalize(x_str, y2)

        chosen = y2
        next_state, _, done, _ = env.step(chosen)

        portfolio_return = (env.balance + env.shares * float(env.df.iloc[env.current_step]['DJIA_Close']) - env.initial_balance) / env.initial_balance
        print(f"Day {env.current_step-1:4d} | r1={r1:+.4f}  r2={r2:+.4f}  Stored={r2>agent.tau}  Return={portfolio_return*100:+6.2f}%")

        returns.append(portfolio_return)
        final_return = returns[-1] if returns else 0
    print(f"\n=== ERL with Qwen2.5-7B FINISHED ===\nFinal Portfolio Return: {final_return*100:+.2f}%")
    print(f"Memory size: {len(agent.memory.reflections)} | Internalized: {len(agent.internalized)}")
    return final_return

torch.cuda.empty_cache()
gc.collect()

EMBED_MODEL = "sentence-transformers/all-MiniLM-L6-v2"
LLM_MODEL = "Qwen/Qwen2.5-7B-Instruct"  

embedder = SentenceTransformer(EMBED_MODEL)
tokenizer = AutoTokenizer.from_pretrained(LLM_MODEL)
model = AutoModelForCausalLM.from_pretrained(
    LLM_MODEL,
    torch_dtype=torch.float16,
    device_map="auto",
    trust_remote_code=True
)
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)

df_news = pd.read_csv('/kaggle/input/datasets/junaidgpu/stock-data-with-financial-news-analysis/Combined_News.csv', on_bad_lines='skip')
df_news['Date'] = pd.to_datetime(df_news['Date'], errors='coerce').dt.date

raw = yf.download('^DJI', start='2008-01-01', end=datetime.now().strftime('%Y-%m-%d'), progress=False)
prices = raw['Close'].reset_index()
prices.columns = ['Date', 'DJIA_Close']
prices['Date'] = pd.to_datetime(prices['Date']).dt.date
df = df_news.merge(prices, on='Date', how='inner')
df = df.dropna().sort_values('Date').reset_index(drop=True)

df = df[df['DJIA_Close'].diff() != 0].reset_index(drop=True)        
df = df[df['Date'] >= pd.to_datetime('2015-01-01').date()].reset_index(drop=True)  

print(f"Using {len(df)} trading days with real price changes.")

env = TradingEnv(df)
agent = ERLAgent(generator, embedder, tokenizer)

run_erl(env, agent, num_days=80)   
```
